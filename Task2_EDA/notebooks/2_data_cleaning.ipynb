### **Cleaning locations Table**
- Collapsed to 19,015 rows, one row per unique zip code prefix
- geolocation_lat/lng averaged across duplicate readings for the same zip
- geolocation_city/state taken via 'first'
- Result stored as locations_cleaned

```python
locations.head()
```

```python
locations.duplicated().sum()
```

```python
locations_cleaned=locations.groupby('geolocation_zip_code_prefix').agg({'geolocation_lat': 'mean', 'geolocation_lng': 'mean','geolocation_city': 'first','geolocation_state': 'first'}).reset_index()  
```
